*Sample data is synthetic and for demonstration only.*

```python
import pandas as pd

df = pd.read_csv('/kaggle/input/blockchain-technology-benchmarks/blockchain_transaction_benchmarks.csv')
df.head()
```

## Average signal scores by network

```python
signals = ['address_validity','hash_validity','confirmation_depth','fee_efficiency','settlement_speed','data_completeness','overall_score']
df.groupby('network')[signals].mean().round(1)
```

## Status distribution

```python
df.groupby(['network','status']).size().unstack(fill_value=0)
```

## Most common weakest signal

```python
df['weakest_signal'].value_counts()
```

## Fee as % of amount

```python
df['fee_pct'] = (df['fee_usdt'] / df['amount_usdt'] * 100).round(2)
df.groupby('network')['fee_pct'].describe().round(2)
```
